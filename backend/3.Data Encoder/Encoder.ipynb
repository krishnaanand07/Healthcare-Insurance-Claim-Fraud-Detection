```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_excel(r'D:\1.Tekworks\Project\Healthcare Insurance Claim\Backend\1.Dataset\Clean data\clean_data.xlsx')
```

```python
data.head()
```

```python
boolean_columns = [
    'Claim_Submitted_Late',
    'Is_Fraudulent'
]
data['Claim_Submitted_Late'] = data['Claim_Submitted_Late'].astype(int)

data['Is_Fraudulent'] = data['Is_Fraudulent'].astype(int)
```

```python
data.head()
```

```python
from sklearn.preprocessing import LabelEncoder
label_columns = [
    'Patient_Gender',
    'Admission_Type',
    'Service_Type'
]

le = LabelEncoder()

for col in label_columns:
    if col in data.columns:
        data[col] = le.fit_transform(data[col].astype(str))
    else:
        print(f'Column {col!r} not found in dataframe; skipping')
```

```python
data.head()
```

```python
data.info()
```

```python
onehot_columns = [
    'Provider_Type',
    'Provider_Specialty',
    'Discharge_Type'
]
existing_onehot = [c for c in onehot_columns if c in data.columns]
if existing_onehot:
    data = pd.get_dummies(data, columns=existing_onehot, drop_first=True)
else:
    print('No one-hot columns found in dataframe; skipping get_dummies')
```

```python
data.head()
```

```python
high_cardinality_columns = [
    'Patient_City',
    'Provider_City',
    'Diagnosis_Code',
    'Procedure_Code'
]
for col in high_cardinality_columns:
    if col in data.columns:
        freq = data[col].value_counts()
        data[col] = data[col].map(freq)
    else:
        print(f'Column {col!r} not found; skipping')
```

```python
data.head()
```

```python
data['Claim_Year'] = data['Claim_Date'].dt.year
data['Claim_Month'] = data['Claim_Date'].dt.month

data['Claim_Delay_Days'] = (
    data['Claim_Date'] - data['Service_Date']
).dt.days
```

```python
data.drop(columns=[
    'Claim_Date',
    'Service_Date',
    'Policy_Expiration_Date'
], inplace=True)
```

```python
data.head(10)
```

```python
bool_cols = data.select_dtypes(include='bool').columns

data[bool_cols] = data[bool_cols].astype(int)
```

```python
data.head(5)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

if 'Patient_State' in data.columns:
    data['Patient_State'] = le.fit_transform(data['Patient_State'].astype(str))
else:
    print('Column Patient_State not found; skipping')

if 'Provider_State' in data.columns:
    data['Provider_State'] = le.fit_transform(data['Provider_State'].astype(str))
else:
    print('Column Provider_State not found; skipping')
```

```python
data.head()
```

```python
data['Is_Fraudulent'].value_counts()
```

```python
# Separate Features and Target
X = data.drop('Is_Fraudulent', axis=1)
y = data['Is_Fraudulent']

# Apply SMOTE
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_smote, y_smote = smote.fit_resample(X, y)

# Convert back to DataFrame
X_smote = pd.DataFrame(X_smote, columns=X.columns)

# Add target column back
df_balanced = X_smote.copy()
df_balanced['Is_Fraudulent'] = y_smote

# Check balanced output
print(df_balanced['Is_Fraudulent'].value_counts())
```
